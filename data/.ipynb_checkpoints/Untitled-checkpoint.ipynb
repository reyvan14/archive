```python
import json

input_file = 'DISC-Law-SFT-Triplet-QA-released.jsonl'
output_file = '3.jsonl'



with open(input_file, 'r', encoding='utf-8') as infile, open(output_file, 'w', encoding='utf-8') as outfile:
    for line in infile:
        data = json.loads(line)
        new_data = {
            "instruction": data.get("input", "").strip(),
            "input": "",
            "output": data.get("output", "").strip()
        }
        outfile.write(json.dumps(new_data, ensure_ascii=False) + '\n')

print(f"转换完成，输出文件保存为：{output_file}")

```

```python
import json

input_file = 'DISC-Law-SFT-Triplet-QA-released-filter.jsonl'
output_file = '3.jsonl'

with open(input_file, 'r', encoding='utf-8') as infile, open(output_file, 'w', encoding='utf-8') as outfile:
    for line in infile:
        data = json.loads(line)

        # instruction 直接使用 input，不添加模板提示
        instruction = data.get("input", "").strip()

        new_data = {
            "instruction": instruction,
            "input": "",
            "output": data.get("output", "").strip()
        }

        outfile.write(json.dumps(new_data, ensure_ascii=False) + '\n')

print(f"转换完成，共输出到：{output_file}")
```

```python
import json

input_file = 'DISC-Law-SFT-Triplet-released-filter.jsonl'
output_file = '4.jsonl'

with open(input_file, 'r', encoding='utf-8') as infile, open(output_file, 'w', encoding='utf-8') as outfile:
    for line in infile:
        data = json.loads(line)

        # 合并参考法条内容
        references = data.get("reference", [])
        references_text = "适用法条：\n" + "\n".join(references) if references else ""

        # 合并 input 和 reference 到 instruction
        instruction = data.get("input", "").strip()
        if references_text:
            instruction += "\n\n" + references_text

        new_data = {
            "instruction": instruction.strip(),
            "input": "",
            "output": data.get("output", "").strip()
        }

        outfile.write(json.dumps(new_data, ensure_ascii=False) + '\n')

print(f"转换完成，输出文件保存在：{output_file}")
```

```python
import json
import re

def is_mixed_and_messy(text):
    # 同时含有中英文
    has_chinese = re.search(r'[\u4e00-\u9fff]', text)
    has_english = re.search(r'[a-zA-Z]', text)
    if not (has_chinese and has_english):
        return False

    # 判断是否有英文被截断的特征，如换行分词、奇怪的拼接
    messy_patterns = [
        r'[a-zA-Z]\s*\n\s*[a-zA-Z]',   # 英文被换行截断
        r'\n{2,}',                     # 多个换行
        r'[a-zA-Z]\n[^\n]*[\u4e00-\u9fff]',  # 中英混杂断句
    ]
    return any(re.search(p, text) for p in messy_patterns)

input_file = 'DISC-Law-SFT-Triplet-released.jsonl'
output_file = 'DISC-Law-SFT-Triplet-released-filter.jsonl'

count_removed = 0
with open(input_file, 'r', encoding='utf-8') as infile, open(output_file, 'w', encoding='utf-8') as outfile:
    for line in infile:
        data = json.loads(line)
        output_text = data.get("output", "")
        if is_mixed_and_messy(output_text):
            count_removed += 1
            continue  # 跳过这条样本
        # 保留格式处理：将 input 填入 instruction
        new_data = {
            "instruction": data.get("input", "").strip(),
            "input": "",
            "output": output_text.strip()
        }
        outfile.write(json.dumps(new_data, ensure_ascii=False) + '\n')

print(f"清理完成，共移除 {count_removed} 条混乱样本。结果已保存至：{output_file}")
```
